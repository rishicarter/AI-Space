## Byte Pair Encoding
> Algorithm in words
1. Take Input text
2. Convert to char-tokens
3. Find highest freq pairs
4. Merge the pairs
5. DO this N times

```python
i_text = "Cat ate the bat very fast."
```

```python
freq = {}
for i in range(len(i_text)-1):
    if i_text[i] == " " or i_text[i+1] == " ":
        continue
    f = freq.setdefault(i_text[i] + i_text[i+1], 0)
    f += 1
    freq[i_text[i] + i_text[i+1]] = f
bestk, bestv = "", 0
for k, v in freq.items():
    if v > bestv:
        bestk = k
        bestv = v
freq, bestk
```
